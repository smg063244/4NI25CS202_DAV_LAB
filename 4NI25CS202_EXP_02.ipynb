```python
#EDA
import pandas as pd
import numpy as np
data={
    'Name': ['Alice', 'Bob', 'Claire','Amit','Divya','Ravi','kishan','kishan'],
    'Age': [25, 30, 27,20, np.nan,24,25,25],
    'Marks': [87, 90, np.nan,86,67,86,76,76],
    'City': ['Mysuru','Bengaluru','Mysuru',np.nan,'Delhi','Bengaluru','goa','goa']
    }
df=pd.DataFrame(data)
print("DataSet:")
print(df)
```

```python
print("\nShape of dataset (rows, columns):",df.shape)
print("\nColumn names:",df.columns.tolist())
print("\nData types of each column:")
print(df.dtypes)
print("\nDataset info:")
print(df.info())
```

```python
print("\nDescriptive statastics (numerical columns):")
print(df.describe())
# print("\nDescriptive statastics (all columns):")
# print(df.describe(include='all'))
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nMode:\n",df.mode(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
print("\nVariance:\n",df.var(numeric_only=True))
```

```python
#identifying missing values
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values:",df.isnull().sum().sum())
missing_percentage=(df.isnull().sum()/len(df))*100
print("\nPercentage of missing values in each column:")
print(missing_percentage)
```

```python
#identifying duplicates
print("\nDuplicate rows in dataset:")
# print(df.duplicated())
print(df[df.duplicated()])
print("\nTotal duplicate rows:",df.duplicated().sum())
print("\n EDA process completed.")
```

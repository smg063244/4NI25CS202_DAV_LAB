Outlier detection and treatment using IQR and Z-Score

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

z-score=(value-mean)/std

```python
df=pd.read_csv('student_performance_dataset.csv')
print("First 5 rows of the dataset")
print(df.head())
print("\nStatistical summary of the dataset")
print(df.describe())
```

Artificial inserted a few outliers

```python
numeric_columns=['Study_Hours_per_Week','Attendance_Rate','Past_Exam_Scores','Final_Exam_Score']
```

```python
df.loc[0,'Study_Hours_per_Week']=120
df.loc[1,'Attendance_Rate']=250
df.loc[2,'Past_Exam_Scores']=-20
df.loc[3,'Final_Exam_Score']=200
print("\nInserted a few aritificial outlier values")
```

Outlier Detection using IQR method

```python
print("\n---IQR method outlier detection---")
for col in numeric_columns:
  Q1=df[col].quantile(0.25)
  Q2=df[col].quantile(0.75)
  IQR=Q2-Q1
  lower=Q1-1.5*IQR
  upper=Q2+1.5*IQR
  outliers=df[(df[col]<lower) | (df[col]>upper)]
  print(f"{col}:{len(outliers)} outlier(s) found (limits:{lower:.2f} to {upper:.2f})")
```

Outlier Detection using Z-score method

```python
print("\n---Z-Score Method---")
for col in numeric_columns:
  z_scores=(df[col]- df[col].mean())/df[col].std()
  outliers=df[np.abs(z_scores)>3]
  print(f"{col}:{len(outliers)} outlier(s) found (|z| > 3)")
```

Treat Outliers (CAP THEM TO THE IQR LIMITS)

```python
df_treated=df.copy()
print("\n--- Treating Outliers (Capping using IQR limits)")
for col in numeric_columns:
  Q1=df_treated[col].quantile(0.25)
  Q2=df_treated[col].quantile(0.75)
  IQR=Q2-Q1
  lower=Q1-1.5*IQR
  upper=Q2+1.5*IQR

  df_treated[col]=df_treated[col].clip(lower,upper)
  print(f"{col}: Outliers capped between {lower:.2f} and {upper:.2f}")
```

Compare Statistics before and after

```python
print("\nStatastics BEFORE treatment:")
print(df[numeric_columns].describe())
print("\nStatistics AFTER treatment:")
print(df_treated[numeric_columns].describe())
```

Save the cleaned dataset

```python
df_treated.to_csv("student_performance_dataset_cleaned.csv",index=False)
print("\nTreated dataset saved as 'student_performance_dataset_cleaned.csv'")
print("Program executed Successfully")
```

```python
import pandas as pd
import numpy as np
marks =np.array([83,7,55,67,47])
print("Mean:",np.mean(marks))
print("Median:",np.median(marks))
print("Standard Deviation:",np.std(marks))
print("Variance:",np.var(marks))
print("Max:",np.max(marks))
print("Min:",np.min(marks))
print("Sum:",np.sum(marks))
```

```python
data={
    "name": [ "Aisha", "salim","kiran","Suraj","suma"],
    "department" :["CSE","ISE","CSE","AIML","AIML"],
    "marks":[97,86,90,49,50]
}
df=pd.DataFrame(data)
df.to_csv("student_data.csv",index=False)
df.to_json("students.json",orient="records")
```

```python
print("\n--reading csv---")
df1=pd.read_csv("student_data.csv")
df1.head()
print("\n--reading json---")
df2=pd.read_json("students.json")
print(df2)
```

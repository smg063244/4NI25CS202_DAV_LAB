```python
import numpy as np
import pandas as pd
marks=np.array([78,54,56,90,34])
print("Marks: ",marks)
print("Type: ",type(marks),marks.dtype)
print("Number of dimensions: ",marks.ndim)
print("Shape: ",marks.shape)
print("Number of elements: ",marks.size)
```

```python
print("Add 5: ",marks+5)
print("Multiply by 2:",marks*2)
print("Division by 2:",marks/2)
print("Square:",marks**2)
```

```python
print("Sum:",marks.sum())
print("Min:",marks.min())
print("Max:",marks.max())
print("Mean:",marks.mean())
print("Standard deviation:",marks.std())
print("Varience:",marks.var())
```

```python
print("First mark:",marks[0])
print("Second mark:",marks[1])
print("Last mark:",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First 3:",marks[:3])
print("From index 2:",marks[2:])
print("Every second number:",marks[::2])
```

```python
student_marks=np.array([
    [78,56,90],
    [78,98,70],
    [67,90,76]
])
print(student_marks)
print("Shape:",student_marks.shape)
```

```python
print("Mean of all values:",student_marks.mean())
print("Mean of each column:",student_marks.mean(axis=0))
print("Mean of each row:",student_marks.mean(axis=1))
```
